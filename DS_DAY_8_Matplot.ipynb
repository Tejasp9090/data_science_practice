```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
plt
```

```python
state=['Maharashtra','Gujrat','Up','Himachal Pradesh']
record=[140,90,100,400]
```

```python
plt.bar(state,record,width=0.4,color='#663873',label='Covid Count')
plt.title('State vs Records')
plt.xlabel('State from india')
plt.ylabel('Count')
plt.legend()
plt.savefig('state_1.jpeg',dpi=400)
plt.show()
```

```python
plt.figure(figsize=(3,3))
plt.bar(state,record,width=0.4,color='#663873',label='Covid Count')
plt.title('State vs Records')
plt.xlabel('State from india')
plt.ylabel('Count')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(3,3))
plt.plot(state,record,color='#663873',label='Covid Count')
plt.title('State vs Records')
plt.xlabel('State from india')
plt.ylabel('Count')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```
